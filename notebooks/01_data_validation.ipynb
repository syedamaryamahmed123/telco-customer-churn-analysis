```python
df['TotalCharges'] = pd.to_numeric(df['TotalCharges'], errors='coerce')
```

```python
categorical_columns = [
    "gender",
    "Partner",
    "Dependents",
    "PhoneService",
    "MultipleLines",
    "InternetService",
    "OnlineSecurity",
    "OnlineBackup",
    "DeviceProtection",
    "TechSupport",
    "StreamingTV",
    "StreamingMovies",
    "Contract",
    "PaperlessBilling",
    "PaymentMethod",
    "Churn"
]

for column in categorical_columns:
    print(f"\n--- {column} ---")
    print(df[column].value_counts())
```

```python
pd.crosstab(
    df["Churn"],
    df["Churn_Flag"]
)
```

```python
df["Tenure_Group"].value_counts()
```

```python
tenure_order = [
    "0-12 Months",
    "13-24 Months",
    "25-48 Months",
    "49-60 Months",
    "60+ Months"
]

df["Tenure_Group"] = pd.Categorical(
    df["Tenure_Group"],
    categories=tenure_order,
    ordered=True
)
```

```python
df["Tenure_Group"].value_counts().sort_index()
```

```python
charge_order = [
    "Under $30",
    "$30-$59",
    "$60-$89",
    "$90+"
]

df["Monthly_Charges_Group"] = pd.Categorical(
    df["Monthly_Charges_Group"],
    categories=charge_order,
    ordered=True
)

df["Monthly_Charges_Group"].value_counts().sort_index()
```
